```python
customers_data = [
    (101, "Amit Sharma", "Ahmedabad", "2025-01-10"),
    (102, "Neha Patel", "Mumbai", "2025-02-15"),
    (103, "Rahul Shah", "Pune", "2025-03-20"),
    (104, "Priya Mehta", "Delhi", "2025-04-10"),
    (105, "Karan Joshi", "Ahmedabad", "2025-05-12"),
    (106, "Sneha Desai", "Mumbai", "2025-06-18"),
    (107, "Rohan Gupta", "Delhi", "2025-07-22"),
    (108, "Pooja Singh", "Pune", "2025-08-25")
]

customers_schema = """
    CustomerID INT,
    CustomerName STRING,
    City STRING,
    RegistrationDate STRING
"""

customers_df = spark.createDataFrame(
    customers_data,
    customers_schema
)

display(customers_df)
```

```python
products_data = [
    (201, "Laptop", "Electronics", 55000),
    (202, "Mobile", "Electronics", 25000),
    (203, "Keyboard", "Accessories", 1500),
    (204, "Mouse", "Accessories", 800),
    (205, "Office Chair", "Furniture", 7500),
    (206, "Study Table", "Furniture", 12000),
    (207, "Headphones", "Electronics", 3000),
    (208, "Monitor", "Electronics", 15000)
]

products_schema = """
    ProductID INT,
    ProductName STRING,
    Category STRING,
    UnitPrice INT
"""

products_df = spark.createDataFrame(
    products_data,
    products_schema
)

display(products_df)
```

```python
sales_data = [
    (1001, 101, 201, 1, "2026-01-05"),
    (1002, 102, 202, 2, "2026-01-10"),
    (1003, 103, 203, 3, "2026-01-15"),
    (1004, 104, 204, 5, "2026-01-20"),
    (1005, 105, 205, 2, "2026-02-05"),
    (1006, 106, 206, 1, "2026-02-10"),
    (1007, 107, 207, 4, "2026-02-15"),
    (1008, 108, 208, 2, "2026-02-20"),
    (1009, 101, 202, 1, "2026-03-05"),
    (1010, 102, 201, 1, "2026-03-10"),
    (1011, 103, 205, 3, "2026-03-15"),
    (1012, 104, 206, 2, "2026-03-20"),
    (1013, 105, 207, 5, "2026-03-25"),
    (1014, 106, 208, 1, "2026-03-28"),

    # Invalid data
    (1015, 101, 201, -2, "2026-03-29"),
    (1016, None, 202, 1, "2026-03-29"),
    (1017, 999, 203, 2, "2026-03-30"),
    (1018, 102, 999, 1, "2026-03-30"),
    (1019, 103, 204, None, "2026-03-30"),
    (1020, 104, 205, 1, "InvalidDate"),

    # Duplicate record
    (1001, 101, 201, 1, "2026-01-05")
]

sales_schema = """
    SalesID INT,
    CustomerID INT,
    ProductID INT,
    Quantity INT,
    SalesDate STRING
"""

sales_df = spark.createDataFrame(
    sales_data,
    sales_schema
)

display(sales_df)
```

```python
datasets = {
    "customers": customers_df,
    "products": products_df,
    "sales": sales_df
}

for name, dataframe in datasets.items():

    dataframe.coalesce(1) \
        .write \
        .mode("overwrite") \
        .option("header", True) \
        .csv(f"{base_path}/{name}")

    print(f"{name} successfully generated")
```
